```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent   # notebook berjalan dari folder notebooks/, jadi naik satu level ke root
DATA_DIR = ROOT / "data/pasar_tradisional/Cleaned_After_Imputation"

files = sorted(DATA_DIR.glob("*.csv"))
print(len(files), "file ditemukan")
print(files[0].name)
```

```python
df = pd.read_csv(files[0])
print(df.shape)
df.dtypes
```

```python
import re

frames = []
for f in files:
    # "final_komoditas_tradisional_bawang_merah_2022_2026.csv" → "bawang_merah"
    komoditas = re.sub(r"^final_komoditas_tradisional_|_\d{4}_\d{4}$", "", f.stem)

    wide = pd.read_csv(f)
    long = wide.melt(id_vars="Date_Param", var_name="provinsi", value_name="harga")
    long["komoditas"] = komoditas
    frames.append(long)

harga = pd.concat(frames, ignore_index=True)
harga = harga.rename(columns={"Date_Param": "tanggal"})
harga["tanggal"] = pd.to_datetime(harga["tanggal"])   # sesuaikan format kalau Cell 4 menunjukkan dd/mm/yyyy

print(harga.shape)
print(harga["komoditas"].unique())
harga.head()
```

```python
n_hari = harga["tanggal"].nunique()
print("hari:", n_hari, "| rentang:", harga["tanggal"].min().date(), "→", harga["tanggal"].max().date())
print("baris aktual:", len(harga), "| ekspektasi:", n_hari * 34 * 10)
print("duplikat grain:", harga.duplicated(["tanggal", "provinsi", "komoditas"]).sum())
print("harga <= 0:", (harga["harga"] <= 0).sum())
```

```python
# tanggal kalender yang tidak ada sama sekali
semua_hari = pd.date_range(harga["tanggal"].min(), harga["tanggal"].max())
hilang = semua_hari.difference(harga["tanggal"].unique())
print(len(hilang), "tanggal hilang")
print(pd.Series(hilang).dt.strftime("%Y-%m-%d (%a)").tolist())

# 16 kombinasi komoditas-tanggal yang hilang
per_kom = harga.groupby("komoditas")["tanggal"].nunique()
print(per_kom[per_kom < harga["tanggal"].nunique()])
```

## Data Quality Findings
- Source data was pre-imputed (missing cells filled); imputation method unknown → volatility may be understated.
- 87 of 1,504 calendar days (5.8%) absent across all commodities — mostly sporadic weekends,
  plus 2022 national holidays (Nyepi, Idulfitri, Ascension Day, Pancasila Day).
- Cabai merah & cabai rawit have 2 extra dates (1,417 vs 1,415 days) not present for other commodities.
- No duplicates at grain (date × province × commodity); no zero or negative prices.
- Implication: day-over-day changes computed via LAG may span multiple days → trends/volatility analyzed at weekly/monthly level.
